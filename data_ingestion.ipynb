```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df=pd.read_csv(r"E:\Projects\telecom_customer_churn.csv")
```

```python
engine = create_engine("mysql+pymysql://root:surma0210@localhost/telecom_customer_churn_analysis")
```

```python
df.columns = df.columns.str.replace(' ', '_')

df.to_sql('telecom_customer_churn', con=engine, if_exists='replace', index=False)

print(f"Imported {len(df)} rows")
```
